## 3. DATA CLEANING

```python
# 1. Re-read with postal code as string
df = pd.read_csv(
    "../data/raw/raw_ecommerce_sales_customer_analytics_150k.csv",
    dtype={"customer_postal_code": "string"}
)
```

```python
# 2. Parse dates and times
df["order_date"] = pd.to_datetime(df["order_date"])
df["order_time"] = pd.to_datetime(df["order_time"], format="%H:%M:%S").dt.time
```

```python
# 3. Preserve postal code (redundant safety)
df["customer_postal_code"] = df["customer_postal_code"].str.zfill(5)
```

```python
# 4. Delivery delay flag
df["delivery_delay_days"] = df["delivery_days"] - df["estimated_delivery_days"]
df["is_delayed"] = df["delivery_delay_days"] > 0
```

```python
# 5. Status flags
df["is_cancelled"] = df["order_status"].str.lower().eq("cancelled")
df["is_returned"]  = df["order_status"].str.lower().eq("returned")
df["is_pending"]   = df["order_status"].str.lower().eq("pending")
df["is_completed"] = df["order_status"].str.lower().eq("completed")
```

```python
# 6. Age band (for Tableau segmentation)
df["age_band"] = pd.cut(
    df["customer_age"],
    bins=[0, 25, 35, 45, 55, 65, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65", "66+"]
)
```

```python
import pandas as pd

# 7.1 Load ECB data and filter for 2021-2025
ecb_df = pd.read_csv("../data/raw/eurofxref-hist.csv", index_col="Date", parse_dates=True)
ecb_df = ecb_df.sort_index()
ecb_filtered = ecb_df.loc["2021-01-01":"2025-12-31"]

# 7.2 Calculate 2021-2025 averages
currencies_needed = ["USD", "GBP", "AUD", "CAD", "INR"]
ecb_avg = ecb_filtered[currencies_needed].mean().round(4)

# 7.3 Calculate correct cross-rates dynamically
eur_to_usd = ecb_avg["USD"]

fx_to_usd = {
    "USD": 1.0000,
    "EUR": eur_to_usd,  # Fixed: 1 EUR = 1.1060 USD
    "AED": round(1 / 3.6725, 4) # Fixed peg
}

for curr in ["GBP", "AUD", "CAD", "INR"]:
    fx_to_usd[curr] = round((1 / ecb_avg[curr]) * eur_to_usd, 4)

print("Calculated fx_to_usd dictionary:")
print(fx_to_usd)

# 8. Map rates and validate
df["fx_rate_to_usd"] = df["currency"].map(fx_to_usd)

print("\nCurrencies without exchange rates:")
print(df.loc[df["fx_rate_to_usd"].isna(), "currency"].unique())

# This will crash the cell if any currency is missing a rate
assert df["fx_rate_to_usd"].notna().all(), "Missing exchange rates detected!"

# 9. Apply Conversion
cols_to_convert = ["gross_sales", "discount_amount", "tax_amount", "shipping_cost",
                   "net_sales", "product_cost", "profit", "customer_lifetime_value"]

for col in cols_to_convert:
    df[f"{col}_usd"] = (df[col] * df["fx_rate_to_usd"]).round(2)

# 10. Loss-Making Flag
df["is_loss_making"] = df["profit_usd"] < 0

print("\n✅ Currency conversion and validation complete.")
```

```python
# Validation Check
total_rev = df['net_sales_usd'].sum()
reference_rev = 177_134_263.74
match_pct = 100 * total_rev / reference_rev

print("=== VALIDATION ===")
print(f"Rows: {len(df)}")
print(f"Revenue (USD): ${total_rev:,.2f}")
print(f"Reference (stats file): ${reference_rev:,.2f}")
print(f"Match: {match_pct:.1f}%")
print(f"Loss-making orders: {df['is_loss_making'].sum():,}")
print(f"Returned orders: {df['is_returned'].sum():,}")
```

```python
# Save the final clean data
df.to_csv("../data/processed/orders_clean.csv", index=False)
print("\n✅ Saved to data/processed/orders_clean.csv")
```

## LOAD TO POSTGREQSL DATABASE

```python
import pandas as pd
import time
import getpass
from sqlalchemy import create_engine
# 1. Connect to Postgres
# This will pop up a secure box asking for the password when you run it.

pwd = getpass.getpass("Enter PostgreSQL password: ")
db_url = f"postgresql://postgres:{pwd}@localhost:5432/ecommerce_analytics"
engine = create_engine(db_url)
```

```python
# 2. Read the clean CSV
df_clean = pd.read_csv("../data/processed/orders_clean.csv")

# 3. Ensure order_date is a date object so Postgres recognizes it as a DATE column
df_clean["order_date"] = pd.to_datetime(df_clean["order_date"]).dt.date

# 4. Load to Postgres
start_time = time.time()
print("Loading data to PostgreSQL...")
df_clean.to_sql("orders", engine, if_exists="replace", index=False)

print(f"✅ Loaded {len(df_clean)} rows in {time.time() - start_time:.1f} seconds.")
```
